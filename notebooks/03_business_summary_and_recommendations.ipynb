## Project Objective

The goal of this analysis was to examine patterns and drivers of subway delays across divisions, time periods, and operational categories in order to identify high-impact areas for performance improvement.

## Key Findings
### 1. Delay Distribution
Delays are heavily right-skewed, with most observations showing low delay counts and a small number of extreme disruption events.

### 2. Weekday vs Weekend Operations
Weekday delays exhibit higher median values and greater variability compared to weekends, suggesting increased operational complexity during workdays.

### 3. Pandemic Impact
A sharp decline in delays occurred in early 2020, likely reflecting reduced operations during COVID-19. Delays recovered and stabilized in subsequent years.

### 4. Primary Delay Drivers
Infrastructure and equipment issues are the largest contributors to total delays, followed by police/medical incidents and planned maintenance work.

## Business Recommendations
1. Prioritize Infrastructure Investment
   Since infrastructure and equipment issues contribute the most delays, targeted upgrades and preventive maintenance could significantly reduce total disruption.

2. Strengthen Weekday Operational Planning
   Given higher variability and extreme events during weekdays, enhanced staffing and monitoring may reduce peak-day disruptions.

3. Optimize Planned Maintenance Scheduling
   Re-evaluating timing of planned ROW work may minimize passenger impact.

4. Monitor High-Impact Event Patterns
   Further investigation into extreme delay events could improve contingency planning.
